```python
from math import *
from scipy.constants import G
masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22] #masses in kg?

radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195] #radius in km?

names = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto'] 
dictionary = {}

combi = zip(masses, radii, names)

print(f"so bascially, the escape velocity for EACH planet (and the sun) isssss...:")

for (M, R, N) in combi:
    vesc = sqrt((2*G*M)/(1000*R)) #escape velocity in meters per second?
    print(f"{N}: {vesc:.2e} m/s")
    print(f"An approximation for the mass is: {M:.2e} \n")
    dictionary[N] = vesc

print(dictionary)
```

***Question 3.16: Are NumPy methods faster than a for loop?***

```python
import numpy as np
from time import process_time

N = 1000000
x = np.linspace(0, 1, N)

startl = process_time()  #l for loop
Sum = 0.0
for f in x:
      Sum += f

mu = Sum/x.size
Sum = 0.0
for f in x:
    Sum += (f-mu)**2

stdl = np.sqrt(Sum/x.size)
endl = process_time()
loopt = endl - startl #t for time

print(f"The time for the loop is {loopt} seconds")

startn = process_time() # n for numpy
stdn = np.std(x)
endn = process_time()
numpyt =  endn - startn #t for time
print(f"The time for the numpy calculation is {numpyt} seconds")

```

***Question 3.17: Standard deviation using ndarray and in a one-liner***

```python
import numpy as np
def stdev(i):  #making a function that gives the standard deviation
    
    x = np.array(i)
    mean = x.mean()
    
    difference = (x - mean)**2
    variance = np.mean(difference)
    standarddeviation = np.sqrt(variance)
    return standarddeviation
  
xlist = [1, 2, 3, 4, 5, 6, 7, 8, 9]   
print(stdev(xlist))  #nice defined function result


print(np.std(xlist))  #numpy result


print(np.sqrt(np.mean((np.array(xlist)- np.array(xlist).mean())**2)))  #oneliner result
```

***Question 3.18: Least Squares fitting***

***Question 3.22: Plotting a catenary***

```python
import matplotlib.pyplot as plt
from numpy import *


xlist = list(range(-5, 6, 1))
flist = []
for x in xlist:
    f = (e**x + e**-x)/2
    flist.append(f)

x = xlist
ff = flist
plt.plot(x, ff, marker= 'v', color= 'g', label="Data", markersize=5)
plt.xlabel("x")
plt.ylabel("ff")
plt.title("Chain", fontsize = 16, color = "gray")
plt.legend()
plt.show()
```
