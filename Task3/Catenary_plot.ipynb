```python
import matplotlib.pyplot as plt
from numpy import *


xlist = list(range(-5, 6, 1))
flist = []
for x in xlist:
    f = (e**x + e**-x)/2
    flist.append(f)

x = xlist
ff = flist
plt.plot(x, ff, marker= 'v', color= 'g', label="Data", markersize=5)
plt.xlabel("x")
plt.ylabel("ff")
plt.title("Chain", fontsize = 16, color = "gray")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import *
lijst = np.arange(-5, 6, 1)

f = (e**lijst + e**-lijst)/2

plt.plot(lijst, f, marker= 'o', color= 'b', label="Data", markersize=3)
plt.xlabel("Distance")
plt.ylabel("Height")
plt.title("Catenary", fontsize = 16, color = "gray")
plt.legend()
plt.show()
```
