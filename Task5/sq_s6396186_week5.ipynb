***Question 5.9: Exploring artificial image colors***

```python
import imageio
import numpy as np
import matplotlib.pyplot as plt

#Making the normal plot
img = imageio.imread("m101BW.jpg")
plt.figure()
plt.title("Normal plot")
plt.imshow(img, cmap = 'jet')
plt.colorbar()
plt.show()

#Making the contour plot
levels = np.percentile(img, [90,95,99])

plt.figure()
plt.imshow(img, cmap='grey')
plt.contour(img, levels=levels, colors='red')
plt.title("Contour Plot")
#plt.colorbar()
plt.show()

#Making a plot with only the highest 10% intensity
threshold = np.percentile(img, 90)   # 90th percentile
filtered = np.where(img >= threshold, img, 0)
plt.figure()
plt.imshow(filtered, cmap='jet')  #Gives only the top 10% data
plt.title("Top 10% Intensity Pixels")
plt.colorbar()
plt.show()
```

***Question 5.10: Slicing in a multi dimensional array***

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('ngc6946.txt')
print(f"Loaded data shape: {data.shape}")
cube = data.reshape(40, 50, 70)  #reshaping
print(f"Reshaped cube shape: {cube.shape}")

#Getting the z_profile

z_profile = cube[:, 25, 35]
print(f"Z-profile (middle x,y): ")
print(z_profile)


#Calculating stuff

maximum = z_profile.max()
print(f"The maximum is {maximum}")
minimum = z_profile.min()
print(f"The minimum is {minimum}")
Sum = z_profile.sum()
print(f"The sum is{Sum}")
mean = z_profile.mean()
print(f"The mean is {mean}")
stdev = z_profile.std()
print(f"The standard deviation is {stdev}")

#Creating the plot

plt.plot(z_profile, color="purple")
plt.text(25, 0.06, f"Max: {float(maximum)}")
plt.text(25, 0.055, f"Min: {float(minimum)}")
plt.text(25, 0.05, f"Mean: {float(mean)}")
plt.text(25, 0.045, f"Stdev: {float(stdev):.3f}")
plt.text(25, 0.04, f"Sum: {float(Sum)}")

plt.title("Z-profile")
plt.show()
```

***Question 5.11 : Extract images of a 3D cube***


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('ngc6946.txt')
print(f"Loaded data shape: {data.shape}")
cube = data.reshape(40, 50, 70)  #reshaping
print(f"Reshaped cube shape: {cube.shape}")

image = cube[0, :, :]  #getting the first plane


#Calculating stuff

mean = image.mean()
stdev = image.std()


#Making the plot
levels = [mean, mean + stdev, mean + 2* stdev]
colors = ["red", "blue", "green"]
plt.imshow(image, origin ='lower' , cmap = 'viridis')
plt.colorbar(label = 'intensity')
plt.contour(image, levels=levels, colors=colors)
```

***Question 5.12 : Getting acquainted with classes***

```python
from scipy.constants import c  #in m/s

class Galaxy:    
    """Class for redshift calculation of galaxies"""  
    C = c  
    caH_0 = 3968.5      # Calcium H line in rest frame
    
    def __init__(self, name, calciumHline=caH_0):
        self.name = name
        self.caH = calciumHline
    def redshift(self):        
        z = (self.caH - self.caH_0)/self.caH
        return z
    def velocity(self):
        z = self.redshift()
        v = (z * self.C)/1000
        return v
    def newmethod(self):
        z = self.redshift()
        c = 3.0e8
        v = (c*z)/1000
        return v
    def __str__(self):
        s = f"Galaxy name: {self.name}  Redshift: {self.redshift():.5f}  Velocity: {self.velocity():.5f}"
        return s
        
gal1 = Galaxy("NGC1832", 3994.7)
print(gal1)
```

***Question 5.13: Ellipse plotter with astronomical parameters***
