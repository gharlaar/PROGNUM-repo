***Question 4.3:  Game Making: Rock-Paper-Scissors***

```python
import numpy as np
user = input("Enter R (rock), P (paper) or S (scissors): ")

programme = np.array(["R", "P", "S"])  #Where the computer can choose from
randindx = np.random.randint(0, len(programme), 1) #Gives either R, P or S
bot = programme[randindx] #Couples the programme to the randindx
print(f"The bot gives... {bot}!")

#Conditions and rules of the game

if user == "R" and bot == "S":
    print(f"User wins!")
elif user == bot:
    print(f"It's a tie!")
elif user == "P" and bot == "R":
    print(f"User wins!")
elif user == "S" and bot == "P":
    print(f"User wins!")
elif bot == "R" and user == "S":
    print(f"Bot wins!")
elif bot == "P" and user == "R":
    print(f"Bot wins!")
elif bot == "S" and user == "P":
    print(f"Bot wins!")
```

***Question 4.4: Simulating measurements and statistical analysis***


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import *

mu = 185 #mean in cm
sigma = 8 #standard deviation in cm
datapoints = 100000

height = np.random.normal(mu, sigma, datapoints) #random function

mean = height.mean()
print(f"The mean of this data set is {mean} cm.")

stdev = height.std()
print(f"The standard deviation of this data set is {stdev} cm.")


#Making the plot

plt.axvline(x = mean, color = 'r', label = f"Mean = {mean} cm")
plt.hist(height, bins = 40, color = "lightblue")
plt.axvspan(mean-stdev, mean+stdev, alpha=0.5, color='lightgreen', label = f"Standard deviation = {stdev}")

plt.legend()
plt.show()


# Calculating the probability for smaller than h

h = 180
cdf = norm.cdf(h, scale = stdev, loc = mean)
Probabitlity = cdf*100
print(Probabitlity)
```
