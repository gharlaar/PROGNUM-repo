***Question 4.4: Simulating measurements and statistical analysis***


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import *

mu = 185 #mean in cm
sigma = 8 #standard deviation in cm
datapoints = 100000

height = np.random.normal(mu, sigma, datapoints) #random function

mean = height.mean()
print(f"The mean of this data set is {mean} cm.")

stdev = height.std()
print(f"The standard deviation of this data set is {stdev} cm.")


#Making the plot

plt.axvline(x = mean, color = 'r', label = f"Mean = {mean} cm")
plt.hist(height, bins = 40, color = "lightblue")
plt.axvspan(mean-stdev, mean+stdev, alpha=0.5, color='lightgreen', label = f"Standard deviation = {stdev}")

plt.legend()
plt.show()


# Calculating the probability for smaller than h

h = 180
cdf = norm.cdf(h, scale = stdev, loc = mean)
Probabitlity = cdf*100
print(f"So the probability of the height being smaller or equal to 180 cm is {Probabitlity} percent.")

#Calculating the probabitlity of a height located within one standard deviation range

h_up = mean + stdev
h_down = mean - stdev
cdfup = norm.cdf(h_up, scale = stdev, loc = mean)
cdfdown = norm.cdf(h_down, scale = stdev, loc = mean)
Prob_up = cdfup*100
Prob_down = cdfdown*100

Probability_between = Prob_up - Prob_down

print(f"So the probability of the height being within one standard deviation of the mean is {Probability_between} percent.")
```

***Question 4.6: Integral with Monte Carlo***

```python
import numpy as np
from math import *

#Giving the boundaries

Upper_boundary = float(input(f"Enter the upper boundary here: "))
Lower_boundary = float(input(f"Enter the lower boundary here: "))

#Defining the function

def f(Function, x):
    y = eval(Function)
    return y

Function = input(f"Enter an function here: ") #Giving the function
x = np.random.uniform(Lower_boundary, Upper_boundary, 1000) #Generating a list of values for x

total = sum(f(Function, x)) #Calculating the sum of the values for y

integralvalue = ((Upper_boundary - Lower_boundary)/1000) * total #Calculating the integral
print(f"So the value of the integral is {integralvalue}.")
```

***Question 4.12: Rotate a square***

```python
import numpy as np
from math import *
from matplotlib import pyplot as plt

x = np.array([1, 3, 3, 1, 1])
y = np.array([1, 1, 3, 3, 1])

angle_degrees = 30
angle_radians = np.deg2rad(angle_degrees)  #converting to radians

rotor = cos(angle_radians) + 1j*sin(angle_radians) #making the rotor

position = x + 1j*y

#Defining the new position

rotorposition = rotor * position
xr = rotorposition.real
yr = rotorposition.imag

#Making the plot

plt.plot(x, y, color = "aqua", marker = "o")
plt.plot(xr, yr, color = "purple", marker = "o")
plt.gca().set_aspect(1)
plt.grid(color = "yellow")
plt.show()
```

***Question 4.14: Passing functions as arguments***

```python
import numpy as np
from math import *

n = np.array([-100, -10, 0, 10, 100])
f = np.log10

def funcfilter(func, seq):

    
    if type(seq) == int or type(seq) == float:
        seq = [seq]

    new = []
    for i in seq:  #otherwise with one number will give an error over 0D
        if i > 0:
            new.append(i)

    newarray = np.asarray(new)
    result = func(newarray)
    return result

print(funcfilter(f, n))

result_single = funcfilter(f, 100)
print(result_single)
```

***Question 4.16: Calculating Integration using quad()***

```python
from astropy.cosmology import Planck18
import scipy



# Finding the values of constants

H0 = Planck18.H0.value  #km/s/Mpc
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0


def t(z):  #Defining the function 
    H = H0 * sqrt(Om0 * (1+z)**3 + Ode0)
    y = 1/((1+z) * H)  # (Mpc*s)/km
    return y


# Calculating the integral

z = float(input("Enter a value for z: "))
tL, err = scipy.integrate.quad(t, 0, z)
print(tL) 

#Converting to Gyr


Mpc = 3.0857e19  #km
New = tL*Mpc  #gives s**-1
Gyr = 315576e11 #seconds
t_Gyr = New/Gyr
print(t_Gyr)
```

```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.stats import gaussian_kde

# Ranges for x

mean_x = 0
stdev_x = 1
datapoints_x = 10000

x = np.random.normal(mean_x, stdev_x, datapoints_x) #random function


#Ranges for y

mean_y = 2
stdev_y = 1
datapoints_y = 10000

y = np.random.normal(mean_y, stdev_y, datapoints_y) #random function

#Plotting the contour lines
#Grid for evaluating density
data = np.vstack([x, y])
kde = gaussian_kde(data)

xmin, xmax = x.min(), x.max()
ymin, ymax = y.min(), y.max()
xx, yy = np.mgrid[xmin:xmax:200j, ymin:ymax:200j]
positions = np.vstack([xx.ravel(), yy.ravel()])
density = kde(positions).reshape(xx.shape)

levels = np.percentile(density, [0.3, 5, 32]) #Contour levels for the sigma regions

CS = plt.contour(xx, yy, density, levels=levels, colors = 'red') #Draw contours

plt.clabel(CS, fmt = {levels[0]: r'$ 3\sigma$', levels[1]:r'$ 2\sigma$', levels[2]: r'$ 1\sigma$'})
#Plotting a 2D (3D) histogram
plt.gca().set_aspect('equal', adjustable = 'box')
plt.hist2d(x, y, bins = 30, cmap = 'Blues')

plt.xlim(-7, 7)
plt.ylim(-7, 7)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('2D histogram + contour')
#plt.legend() not used for 2D, since there are no labeled plot elements to display
plt.show() 
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------
# Generate Gaussian data
# ---------------------------


N = 50000

mean_x = 0
mean_y = 2

sigma_x = 1
sigma_y = 1

x = np.random.normal(mean_x, sigma_x, N)
y = np.random.normal(mean_y, sigma_y, N)

# ---------------------------
# Plot 2D histogram
# ---------------------------
plt.figure(figsize=(8,6))

plt.hist2d(x, y, bins=30, cmap="Blues")

cbar = plt.colorbar()
cbar.set_label("Counts")

# ---------------------------
# Create grid for contours
# ---------------------------
xgrid = np.linspace(-4,4,200)
ygrid = np.linspace(-2,6,200)

X, Y = np.meshgrid(xgrid, ygrid)

# Gaussian PDF
Z = np.exp(-((X-mean_x)**2/(2*sigma_x**2) +
             (Y-mean_y)**2/(2*sigma_y**2)))

# Sigma levels
levels = [
    np.exp(-0.5*3**2),  # 3σ
    np.exp(-0.5*2**2),  # 2σ
    np.exp(-0.5*1**2)   # 1σ
]

# Plot contours
CS = plt.contour(X, Y, Z,
                 levels=levels,
                 colors="red",
                 linewidths=2)

plt.clabel(CS, inline=True,
           fmt={levels[2]:"1σ", levels[1]:"2σ", levels[0]:"3σ"})

# ---------------------------
# Styling
# ---------------------------
plt.title("2D Histogram + Contour")
plt.xlabel("X")
plt.ylabel("Y")

plt.grid(True, linestyle="--", color="black", alpha=0.6)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
N = 100000
mean_x, mean_y = 0, 2
sigma_x, sigma_y = 1, 1
bins = 30

# Generate Gaussian samples
x = np.random.normal(mean_x, sigma_x, N)
y = np.random.normal(mean_y, sigma_y, N)

# Compute 2D histogram
H, xedges, yedges = np.histogram2d(x, y, bins=bins, density=True)

# Bin centers
xcenters = 0.5 * (xedges[:-1] + xedges[1:])
ycenters = 0.5 * (yedges[:-1] + yedges[1:])
X, Y = np.meshgrid(xcenters, ycenters)

# Normalize histogram to probability
pdf = H / H.sum()

# Flatten and sort bins by probability density
pdf_flat = pdf.flatten()
idx = np.argsort(pdf_flat)[::-1]
pdf_sorted = pdf_flat[idx]

# Cumulative probability
cdf = np.cumsum(pdf_sorted)

# Sigma probability levels (for Gaussian)
sigma_probs = [0.6827, 0.9545, 0.9973]

# Find density thresholds corresponding to these probabilities
levels = []
for p in sigma_probs:
    levels.append(pdf_sorted[np.searchsorted(cdf, p)])

levels = np.sort(levels)

# Plot
plt.figure(figsize=(7,6))

# 2D histogram
hist = plt.hist2d(x, y, bins=bins, cmap='Blues')
plt.colorbar(label='Counts')

# Contours (note transpose to match orientation)
CS = plt.contour(X, Y, pdf.T, levels=levels, colors='red', linewidths=2)
plt.clabel(CS, inline=True,
           fmt={levels[2]:"1σ", levels[1]:"2σ", levels[0]:"3σ"})

plt.xlabel('x')
plt.ylabel('y')
plt.title('2D Histogram with 1σ, 2σ, 3σ Contours')

plt.show()
```
